```python
from Bio import SeqIO

record = SeqIO.read("sequence.fasta", "fasta")
print(record)
```

```python
record.seq.find("ATG")
```

```python
coding_seq = record.seq[59:392]
protein_v1 = coding_seq.translate()
print(protein_v1)
```

```python
coding_seq= record.seq[59:392]
protein_v2 = coding_seq.translate(to_stop=True)
print(protein_v2)
```
